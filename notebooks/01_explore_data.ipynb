# 01 · Explore the data by hand (no AI)
Before trusting an agent, know the answers yourself. The numbers here become the `expected` column in `evals/questions.csv`.

```python
from energy_agent.config import DATA_FILE
from energy_agent.data import load_smard_csv, make_database

df = load_smard_csv(DATA_FILE)
con = make_database(df)
df.head()
```

## 1. Average price over the whole period

```python
con.execute('SELECT round(avg(deutschland_luxemburg), 2) AS avg_price FROM data').df()
```

## 2. Highest hourly price and when

```python
con.execute('SELECT ts, deutschland_luxemburg FROM data ORDER BY deutschland_luxemburg DESC LIMIT 1').df()
```

## 3. Hours with negative prices

```python
con.execute('SELECT count(*) AS negative_hours FROM data WHERE deutschland_luxemburg < 0').df()
```

## 4. Cheapest hour of the day

```python
con.execute('SELECT hour(ts) AS hour, round(avg(deutschland_luxemburg), 2) AS avg_price FROM data GROUP BY 1 ORDER BY 2 LIMIT 3').df()
```

## 5. Month with the highest average price

```python
con.execute("SELECT strftime(ts, '%Y-%m') AS month, round(avg(deutschland_luxemburg), 2) AS avg_price FROM data GROUP BY 1 ORDER BY 2 DESC").df()
```
