## RAG: what does retrieval return?

## The agent
Watch which tools it chooses for a number question vs. a 'why' question.

```python
import logging
logging.basicConfig(level=logging.INFO, format='  %(message)s', force=True)

from energy_agent.agent import ask, build_agent, get_checkpointer

agent = build_agent(toolbox.as_langchain_tools(), checkpointer=get_checkpointer())
ask(agent, 'How many hours had negative prices?', thread_id='notebook')
```

```python
ask(agent, 'Why does that happen?', thread_id='notebook')
```

## Memory
The second question says *that* — the agent only understands it because the checkpointer stored the first turn under the same `thread_id`. Try the same follow-up with a new thread_id and compare.

```python
ask(agent, 'Why does that happen?', thread_id='fresh-thread')
```
